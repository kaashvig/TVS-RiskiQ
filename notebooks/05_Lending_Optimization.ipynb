# 05_Lending_Optimization

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/risk_scored_dataset.csv')
```

```python
def recommend_ltv(risk):

    if risk == "Low":
        return 85

    elif risk == "Medium":
        return 75

    elif risk == "High":
        return 65

    return 50

df["Recommended_LTV"] = df["Risk_Band"].apply(recommend_ltv)
```

```python
df["PD"] = df["Risk_Band"].map({
    "Low": 0.03,
    "Medium": 0.08,
    "High": 0.18,
    "Critical": 0.35
})
```

```python
df["Expected_Loss"] = (
    df["PD"] *
    (
        df["Loan Amount"]
        -
        df["Residual_Value_Forecast"]
    )
)
```

```python
df["Interest_Income"] = (
    df["Loan Amount"]
    *
    df["Cust Net IRR"]
    / 100
    *
    df["Tenure"]
    / 12
)
```

```python
df["Expected_Profit"] = (
    df["Interest_Income"]
    -
    df["Expected_Loss"]
)
```

```python
def recommend_tenure(risk):

    if risk == "Low":
        return 60

    elif risk == "Medium":
        return 48

    elif risk == "High":
        return 36

    return 24

df["Recommended_Tenure"] = (
    df["Risk_Band"]
    .apply(recommend_tenure)
)
```

```python
print("Average Profit")
print(df["Expected_Profit"].mean())

print("Total Portfolio Profit")
print(df["Expected_Profit"].sum())

print("Average Recommended LTV")
print(df["Recommended_LTV"].mean())
```

```python
df.to_csv(
    "../data/processed/lending_optimized_dataset.csv",
    index=False
)
```
